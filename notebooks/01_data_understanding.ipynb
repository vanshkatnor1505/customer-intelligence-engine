## Table Understanding

### 1. Customers
- One row represents: customer ids, and their address info.
- Important columns: customer_city, customer_id.
- Observations: we can get info about customer's state, city. 

### 2. Geolocation
- One row represents: information about zipcode and city, state.
- Important columns: zipcode, city, state.
- Observations: contains longitude and latitude for each zipcode, zipcode represents city and state that can be helpfull.

### 3. Order Items
- One row represents: order dtails along with product and seller details and price of the product , freight value and complte order cycle.
- Important columns: price,  freight value, order_id.
- Observations: contains info about price of productas and freight value.

### 4. Order Payments
- One row represents: contains info about payment details and installments.
- Important columns: payment_type, payment_value, payment_installmets.
- Observations: dataset contains info about payment linked to order id.

### 5. Order Reviews
- One row represents: contains info about reviews and ratings by customers.
- Important columns: review_score
- Observations: review_tittle, review_comment contains to much missing values. review score is looking nice for customer behaviour understanding .

### 6. Orders
- One row represents: gives info about order status.
- Important columns: order_status
- Observations: contains one inportant column order_status whic can correlate to review.

### 7. Products
- One row represents: gives detiled info about the product and its properties.
- Important columns: product_category_name
- Observations: contains product_category_name columns which can tell us about most sold category.

### 8. Sellers
- One row represents: contains info about seller info 
- Important columns: seller_city.
- Observations: we can find which city has sellers with highest sales.

### 9. Category Translation
- One row represents: gives us translated version of prduct categories to english.
- Important columns: product_category_name_english
- Observations: helps to make data more understandable.
